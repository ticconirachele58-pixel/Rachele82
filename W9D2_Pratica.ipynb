```python
stringa = "Python"
lunghezza = len(stringa)
if lunghezza >= 6:
    print(stringa[:3] + "..." + stringa[-3:])
elif lunghezza == 5:
    print(stringa[:3] + "..." + stringa[-2:])
elif lunghezza == 4:
    print(stringa[:3] + "..." + stringa[-1:])
elif lunghezza == 3:
    print(stringa)
elif lunghezza == 2:
    print(stringa)
elif lunghezza == 1:
    print(stringa)
else:
    print("Stringa vuota!")
```

```python
n = 120
fattori = []
d = 2

while d <= n:
    if n % d == 0:
        fattori.append(d)
        n = n // d
    else:
        d += 1

print(fattori)
```

```python
guadagni = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50]
somma = 0

for guadagno in guadagni:
    somma += guadagno

media = somma / len(guadagni)
print(f"La media è: {media}")
```
