```python
print(df.info())
print(df.describe())
age_20_29 = df[(df["Age (years)"] >= 20) & (df["Age (years)"] <= 29)]
age_30_39 = df[(df["Age (years)"] >= 30) & (df["Age (years)"] <= 39)]
age_40_50 = df[(df["Age (years)"] >= 40) & (df["Age (years)"] <= 50)]
```

```python
print("Media pressione diastolica 20-29:", 
      age_20_29["Diastolic blood pressure (mm Hg)"].mean())

print("Media pressione diastolica 30-39:", 
      age_30_39["Diastolic blood pressure (mm Hg)"].mean())

print("Media pressione diastolica 40-50:", 
      age_40_50["Diastolic blood pressure (mm Hg)"].mean())
```
