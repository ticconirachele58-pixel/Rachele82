```python
import pandas as pd

vendite = pd.Series([2000, 5500, 7200, 4320, 1280, 1900,
                     2500, 3900, 6410, 8150, 7100, 5350])

print("max:", vendite.max())
print("min:", vendite.min())
print(">5000:", vendite[vendite > 5000].tolist(), "count:", (vendite > 5000).sum())
print("<3000:", vendite[vendite < 3000].tolist())
print("media:", vendite.mean())
```

```python
La differenza tra un dizionario e una Series è che il dizionario è una struttura dati base di Python che associa chiavi a valori e serve principalmente 
per memorizzare informazioni, mentre la Series di Pandas è una struttura pensata per l’analisi dei dati. Entrambi permettono di accedere ai valori tramite
una chiave, ad esempio `fatturati_dict[1997]` e `fatturati_series[1997]`, ma la Series offre molte funzionalità in più: permette di fare operazioni 
statistiche come media, massimo, minimo e somma in modo diretto, oltre a poter filtrare i dati con condizioni (ad esempio valori maggiori di un certo 
    numero). Con un dizionario queste operazioni non sono disponibili direttamente e andrebbero scritte manualmente con cicli e codice aggiuntivo.
```

```python
import pandas as pd

stipendi = pd.Series([100, 200, 300, 400, 500,
                      600, 700, 800, 900, 1000])
nuovi_stipendi = [5]

for s in stipendi:
    nuovi_stipendi.append(s * 2)

nuovi_stipendi = pd.Series(nuovi_stipendi)

print(nuovi_stipendi)
#esercizio svolto con for
```
